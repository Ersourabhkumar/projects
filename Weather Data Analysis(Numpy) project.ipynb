```python
#1️⃣ Dataset Creation

#We create 365 values for max and min temperature using NumPy’s randint
np.random.seed(42) #lock or fix our data
max_temp=np.random.randint(20,46,365)
min_temp=np.random.randint(5,31,365)
```

```python
# Average Daily Temperature
```

```python
avg_temp=(max_temp+min_temp)/2
```

```python
# Hottest and Coldest Day
hottest=max_temp.max()
coldest=min_temp.min()
```

```python
# Count how many days crossed 40°C
hot_days=np.sum(max_temp>40)
hot_days
```

```python
# Calculate month-wise average temperature
month_days = np.array([31,28,31,30,31,30,31,31,30,31,30,31])

indices = np.cumsum(month_days)

monthly_avg = [
    avg_temp[start:end].mean()
    for start, end in zip([0]+list(indices[:-1]), indices)
]
```

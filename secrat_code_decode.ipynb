```python
import random
import string
def random_str(length=3):
  return "".join(random.choices(string.ascii_lowercase,k=length))
```

```python
def encode_word(word):
  if len(word)>=3:
    modified_word=word[1:]+word[0]
    start=random_str()
    end=random_str()
    return start+modified_word+end
  else:
    return word[::-1]
```

```python
def decode_word(word):
  if len(word)<3:
    word[::-1]
  else:
    strip=word[3:-3]
    original=strip[-1]+strip[:-1]
    return original
```

```python
def translate_message(message, mode='encode'):
    words = message.split()
    translated_words = []

    for word in words:
        if mode == 'encode':
            translated_words.append(encode_word(word))
        elif mode == 'decode':
            translated_words.append(decode_word(word))

    return " ".join(translated_words)
```

```python
import random
import string

def generate_random_chars(length=3):
    """Generates a string of random lowercase letters."""
    return ''.join(random.choices(string.ascii_lowercase, k=length))

def encode_word(word):
    # Rule 1: If the word contains at least 3 characters
    if len(word) >= 3:
        # Remove the first letter and append it at the end
        modified_word = word[1:] + word[0]
        # Append three random characters at the starting and the end
        start_chars = generate_random_chars(3)
        end_chars = generate_random_chars(3)
        return start_chars + modified_word + end_chars
    # Rule 2: Else (less than 3 characters), simply reverse the string
    else:
        return word[::-1]

def decode_word(word):
    # Rule 1: If the word contains less than 3 characters, reverse it
    if len(word) < 3:
        return word[::-1]
    # Rule 2: Else (it was encoded with the 3+ character rule)
    else:
        # Remove the 3 random characters from the start and the end
        strip_random = word[3:-3]
        # Take the last character and move it back to the front
        original_word = strip_random[-1] + strip_random[:-1]
        return original_word

def translate_message(message, mode='encode'):
    words = message.split()
    translated_words = []

    for word in words:
        if mode == 'encode':
            translated_words.append(encode_word(word))
        elif mode == 'decode':
            translated_words.append(decode_word(word))

    return " ".join(translated_words)

# --- Quick Test ---
if __name__ == "__main__":
    secret_message = "Data Science is fun"
    print(f"Original: {secret_message}")

    # Coding
    coded = translate_message(secret_message, mode='encode')
    print(f"Encoded:  {coded}")

    # Decoding
    decoded = translate_message(coded, mode='decode')
    print(f"Decoded:  {decoded}")
```
